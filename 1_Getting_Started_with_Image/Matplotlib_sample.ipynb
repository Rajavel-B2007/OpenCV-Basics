Opening image using opencvis always in BGR format. So there is conversion reqiured for the original image

```python
import  numpy as np
import cv2
from matplotlib import pyplot as plt
# OPEN IMAGE USING OPENCV
img=cv2.imread('dingu.jpg',1)
plt.imshow(img)
plt.show()

# CONVERTING THE OPENCV IMAGE INTO ORIGINAL IMAGE
# METHOD 1:
rev1=cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
plt.imshow(rev1)
plt.show()

#METHOD 2:

rev2=img[:,:,::-1]
plt.imshow(rev2)
plt.show()
```

Opening image using matplotlib is always a original image that is RGB format

```python
# OPENING THE IMAGE ONLY BY MATPLOTLIB

import matplotlib.image as mimg

i=mimg.imread("dingu.jpg")
plt.imshow(i)
plt.colorbar()
plt.show()
```
